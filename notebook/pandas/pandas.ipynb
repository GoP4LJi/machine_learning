## 38. Advanced Sales Analysis


```python
# Monthly revenue and growth.
txn = pd.DataFrame({
    "Date": ["2024-01-05", "2024-02-05", "2024-03-05", "2024-03-10"],
    "Customer": ["A", "A", "B", "C"],
    "Revenue": [1000, 1500, 3000, 2000]
})

txn["Date"] = pd.to_datetime(txn["Date"])
txn["Month"] = txn["Date"].dt.to_period("M")
month_rev = txn.groupby("Month")["Revenue"].sum().reset_index()
month_rev["Growth"] = month_rev["Revenue"].pct_change() * 100
month_rev
```

## 39. Customer Segmentation


```python
# RFM table.
data = pd.DataFrame({
    "Customer": ["A", "A", "B", "C", "C"],
    "Date": ["2024-01-10", "2024-03-10", "2024-02-05", "2024-01-20", "2024-03-15"],
    "Amount": [1000, 1500, 800, 500, 700]
})

data["Date"] = pd.to_datetime(data["Date"])
today = data["Date"].max()
rfm = data.groupby("Customer").agg(
    Recency=("Date", lambda x: (today - x.max()).days),
    Frequency=("Customer", "count"),
    Monetary=("Amount", "sum")
)
rfm
```

## 42. Missing Data Strategy


```python
# Fill missing values branch wise.
data = pd.DataFrame({
    "Branch": ["CSE", "CSE", "IT", "IT"],
    "Marks": [80, None, 90, None],
    "City": ["Delhi", None, "Pune", None]
})

print(data.isnull().mean() * 100)

data["Marks"] = data.groupby("Branch")["Marks"].transform(lambda x: x.fillna(x.median()))
data["City"] = data.groupby("Branch")["City"].transform(lambda x: x.fillna(x.mode()[0]))
data
```

## 43. Window-Based Analysis


```python
# Previous salary and growth.
sal = pd.DataFrame({
    "Name": ["Amit", "Amit", "Priya", "Priya"],
    "Year": [2023, 2024, 2023, 2024],
    "Salary": [40000, 50000, 60000, 66000]
})

sal["Previous"] = sal.groupby("Name")["Salary"].shift(1)
sal["Growth"] = (sal["Salary"] - sal["Previous"]) / sal["Previous"] * 100
sal[sal["Growth"] > 20]
```

## 44. Advanced Ranking


```python
# Total, average and rank branch wise.
score = pd.DataFrame({
    "Name": ["A", "B", "C", "D"],
    "Branch": ["CSE", "CSE", "IT", "IT"],
    "Python": [80, 90, 70, 95],
    "SQL": [85, 88, 75, 92]
})

score["Total"] = score["Python"] + score["SQL"]
score["Average"] = score[["Python", "SQL"]].mean(axis=1)
score["Rank"] = score.groupby("Branch")["Total"].rank(ascending=False).astype(int)
score
```

## 45. Cohort Analysis


```python
# Simple cohort table.
buy = pd.DataFrame({
    "Customer": ["A", "A", "B", "C", "C"],
    "Date": ["2024-01-10", "2024-02-10", "2024-01-20", "2024-02-05", "2024-03-05"]
})

buy["Date"] = pd.to_datetime(buy["Date"])
buy["Month"] = buy["Date"].dt.to_period("M")
buy["Cohort"] = buy.groupby("Customer")["Month"].transform("min")

buy.groupby(["Cohort", "Month"])["Customer"].nunique()
```

## 46. Fraud Detection with Pandas


```python
# Find high amount and quick transactions.
bank = pd.DataFrame({
    "Account": ["A", "A", "A", "B", "B"],
    "Time": pd.to_datetime(["2024-01-01 10:00", "2024-01-01 10:03", "2024-01-02 11:00", "2024-01-01 09:00", "2024-01-01 09:04"]),
    "Amount": [500, 600, 5000, 700, 750]
})

bank["Avg"] = bank.groupby("Account")["Amount"].transform("mean")
high = bank[bank["Amount"] > bank["Avg"] * 2]
bank["Diff"] = bank.groupby("Account")["Time"].diff().dt.total_seconds() / 60
quick = bank[bank["Diff"] <= 5]

print(high)
quick
```

## 47. Inventory Management


```python
# Stock and reorder level.
inv = pd.DataFrame({
    "Product": ["Pen", "Book", "Bag"],
    "Opening": [100, 50, 30],
    "Purchase": [50, 20, 10],
    "Sold": [120, 30, 35],
    "Reorder": [40, 25, 10]
})

inv["Current_Stock"] = inv["Opening"] + inv["Purchase"] - inv["Sold"]
inv["Turnover"] = inv["Sold"] / inv["Opening"]
inv[inv["Current_Stock"] < inv["Reorder"]]
```

## 48. Large Dataset Optimization


```python
# Memory before and after datatype change.
big = pd.DataFrame({
    "Branch": ["CSE", "IT", "CSE", "ECE"] * 1000,
    "Marks": [80, 90, 70, 85] * 1000
})

before = big.memory_usage(deep=True).sum()
big["Branch"] = big["Branch"].astype("category")
big["Marks"] = big["Marks"].astype("int8")
after = big.memory_usage(deep=True).sum()

print("Before:", before)
print("After:", after)
```

## 49. Real-World Data Cleaning Challenge


```python
# Full cleaning example.
raw = pd.DataFrame({
    "Name": [" amit ", "AMIT", "priya", None],
    "Date": ["2024-01-01", "2024-01-01", "wrong", "2024-02-01"],
    "Category": [" cse", "CSE", "it", "IT"],
    "Marks": [80, "80", 999, None]
})

clean = raw.drop_duplicates().copy()
clean["Name"] = clean["Name"].str.strip().str.title().fillna("Unknown")
clean["Category"] = clean["Category"].str.strip().str.upper()
clean["Date"] = pd.to_datetime(clean["Date"], errors="coerce")
clean["Marks"] = pd.to_numeric(clean["Marks"], errors="coerce")
clean["Marks"] = clean["Marks"].fillna(clean["Marks"].median())
clean = clean[clean["Marks"] <= 100]
clean
```

## 50. Capstone Pandas Project - E-Commerce Analytics


```python
# Small ecommerce analysis.
customers = pd.DataFrame({"Customer_ID": [1,2,3], "Name": ["Amit", "Priya", "Neha"], "Region": ["North", "South", "North"]})
products = pd.DataFrame({"Product_ID": [11,12], "Product": ["Laptop", "Mobile"], "Price": [50000, 20000]})
orders = pd.DataFrame({
    "Order_ID": [101,102,103,104],
    "Customer_ID": [1,2,1,3],
    "Product_ID": [11,12,12,11],
    "Qty": [1,2,1,1],
    "Date": ["2024-01-10", "2024-01-20", "2024-02-05", "2024-02-15"]
})

orders["Date"] = pd.to_datetime(orders["Date"])
final = orders.merge(customers, on="Customer_ID").merge(products, on="Product_ID")
final["Revenue"] = final["Qty"] * final["Price"]
final["Month"] = final["Date"].dt.to_period("M")

monthly_sales = final.groupby("Month")["Revenue"].sum()
product_sales = final.groupby("Product")["Revenue"].sum()
region_sales = final.groupby("Region")["Revenue"].sum()
top_customer = final.groupby("Name")["Revenue"].sum().sort_values(ascending=False)

print(monthly_sales)
print(product_sales)
print(region_sales)
print(top_customer)
print("Final business insights")
print("North region has more sales.")
print("Laptop gives more revenue.")
print("Amit is top customer.")
```
