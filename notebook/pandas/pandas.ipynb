#  BASIC PANDAS EXERCISE

What is Pandas?
pandas is an open-source data analysis and manipulation library for Python. It provides data structures like Series and DataFrame that make it easy to work with structured data, perform data cleaning, transformation, and analysis.


```python
import pandas as pd
import numpy as np
import seaborn as sns
```

#  1. Create a DataFrame

```python
# Q Create a DataFrame containing student names, roll numbers, branch, and marks.
student_data = {
    'Student Name': ['Aarav', 'Ananya', 'Diya', 'Ishaan', 'Kabir', 'Meera', 'Rahul', 'Riya',  'Arjun', 'Saanvi'],
    'Roll Number': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Branch': ['CSE', 'ECE', 'ME', 'CSE', 'IT', 'CE', 'CSE', 'ECE', 'IT', 'ME'],
    'Marks': [85, 92, 78, 65, 88, 74, 91, 83, 79, 81]
}

df = pd.DataFrame(student_data)
```

```python
# Q Display the first 5 and last 5 records.
print(df.head())
print(df.tail())
```

# 2. Series Operations

```python
# Q o	Create a Pandas Series containing marks of 10 students.
marks = student_data["Marks"]
Sname = student_data['Student Name']
S = pd.Series(marks , index= Sname)
S
```

```python
print("Max : ", S.max())
print("Min : ",S.min())
print("Mean : ",S.mean())
print("Median : ",S.median())
```

# 7. Filtering Data

```python
#o	Find all students who scored more than 75 marks.
df[df["Marks"] >= 75]
```

# 8. Multiple Conditions

```python
# o	Find students from the CSE branch who scored more than 80.
df[(df["Marks"]  >= 80)  & (  df["Branch"] == "CSE")]
```

# 9. Adding a Column

```python
## o	Add a Result column based on marks:
# 	Marks ≥ 40 → Pass
# 	Marks < 40 → Fail

df["Result"] = np.where(df["Marks"] >= 40, "Pass", "Fail")
df.head()
```

# 11. Missing Values

```python
data = {
    "Roll Number":[101, 102, np.nan, 104, 105, 106, np.nan, 108, 109, 110],
    "Student Name":["Aarav", "Ananya", "Diya", np.nan, "Kabir", "Meera", "Rahul", "Riya", np.nan, "Saanvi"],
    "Branch":["CSE", np.nan, "ME", "CSE", "IT", "CE", np.nan, "ECE", "IT", "ME"],
    "Marks":[85, np.nan, 78, 65, 88, np.nan, 91, 83, 79, 81],}

df_m = pd.DataFrame(data)
df_m
```

```python
# o	Identify missing values using isnull() and isna().
df_m.isna().sum()
```

# 12. Handling Missing Values

```python
df_m['Marks'] = df_m['Marks'].fillna(df_m['Marks'].mean()) 
df_m
```
