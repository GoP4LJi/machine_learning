```python
# Adding a Column
# Q-9: Add a Result column based on marks:
# Marks ≥ 40 → Pass
# Marks < 40 → Fail
# df['result'] = df['Marks'].apply(lambda x:'Pass' if x>= 40 else 'Fail') # Alternative using numpy 
def check_result(marks):
    if marks>40:
        return("Pass")
    else:
        return("Fail")
df['result'] =  df['Marks'].apply(check_result)
print(df)
```

```python
# GroupBy – Average Marks
# Q-17: Given student data containing Branch and Marks, calculate the average marks for each branch.
avg_marks = df.groupby('Branch')['Student_Marks'].mean().reset_index() #reset_index to get the data in normal datafram format
print(avg_marks)
```

```python
# GroupBy – Multiple Aggregations
# Q-18: Find the minimum, maximum, average, and standard deviation of marks for each branch.
branch_stats = df.groupby('Branch')['Student_Marks'].agg(['min','max','mean','std']).reset_index()
print(branch_stats)
```

```python
# Value Counts
# Q-19: Given a dataset of student branches, count the number of students in each branch.
branch_counts = df['Branch'].value_counts()

print(branch_counts)
```

```python
# Conditional Column
# Q-20: Create a Grade column based on the Marks column.
# A: ≥ 90
# B: 75–89
# C: 60–74
# D: 40–59
# F: < 40
def assign_grade(marks):
    if marks >= 90:
        return 'A'
    elif marks >= 75:
        return 'B'
    elif marks >= 60:
        return 'C'
    elif marks >= 40:
        return 'D'
    else:
        return 'F'
df['Grade'] = df['Student_Marks'].apply(assign_grade)
print(df)
```

```python
# Apply Function
# Q-21: Use the apply() function to calculate the percentage of students based on marks obtained out of 100.
data = {
    'Name': ['Aarav', 'Priya', 'rahul', 'Sneha', 'Karan'],
    'Marks_Obtained': [425, 390, 310, 240, 180]
}
df = pd.DataFrame(data)
def calc_per(Marks):
    return (Marks/500)*100
df['Percentage'] = df['Marks_Obtained'].apply(calc_per)
print(df)
```

```python
# DateTime Operations
# Q-24: Create a DataFrame containing employee joining dates.
# Q-25: Extract year, month, and day from the joining date.
df_emp = pd.read_csv('employee_data.csv')

# Convert Joining_Date to datetime format
df_emp['Joining_Date'] = pd.to_datetime(df_emp['Joining_Date'])
```
