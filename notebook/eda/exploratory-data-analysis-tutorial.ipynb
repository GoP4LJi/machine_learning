# Grouping
Pandas `.groupby()` function is used to split the data into groups based on some criteria. The abstract definition of grouping is to provide a mapping of labels to group names.

To understand the concept, it's better to jump right into the example.

Suppose, we want to see the distribution of `Age=Null` based on `Pclass`. First of all, let's retrieved a copy of DataFrame which `Age=Null`.

````python
null_age = df[df['Age'].isnull()].copy() # Use copy() to create a copy of DataFrame to prevent modification leakage.
````

Then `.groupby()` using `Pclass` and call `.size()` function to get the size of each groups.

It will returns a `Series` showing the number of `Null Age` of each classes.

````python
null_age.groupby('Pclass').size()
````

You can use arithmetic operation into `DataFrame` or `Series` and it will operate per row.
The following is an example how to check the `ratio` between each `Classes` in percentage.

````python
null_age.groupby('Pclass').size() / len(null_age) * 100
````

`.groupby()` is very powerful tool to compute operation on every groups. For example, to find the `Average Fare` of each `Passenger Class`, we can use the following command:

````python
df.groupby('Pclass')['Fare'].mean()
````

Use `.reset_index()` to convert back to DataFrame and reset the index.

Or use `.to_frame()` instead if you want to keep the index.

````python
df.groupby('Pclass')['Fare'].mean().reset_index().rename(columns={'Fare':'Mean_Fare'})
````

````python
df.groupby('Pclass')['Fare'].mean().to_frame(name='Mean_Fare')
````

We can also use multiple columns for grouping.

As we know, `Survived` column consists of `1` or `0`, `Survived` or `Not Survived`. To count the number of survivors we can use `.sum()` function. This function return the sum of the values based on axis, `axis=0` sums vertically (per column) and `axis=1` sums horizontally (per row).

````python
df[['SibSp', 'Parch']].sum(axis=0)
````

````python
df[['SibSp', 'Parch']].sum(axis=1)
````

The next following command is to find `how many people survived` the event based on their `classes` and `sexes`.

````python
df.groupby(['Pclass', 'Sex'])['Survived'].sum().to_frame(name='Survived')
````

Next, we will find out the passenger survivability based on their `Classes` and `Port of Embarkation`.

We can use formula `sum / count` or `mean` to get the probability.

````python
pe_group = df.groupby(['Pclass', 'Embarked'])
````

To run complex commands, we can use `.apply()` and `lambda x:` functions.

Instead of `pe_group['Survived'].mean()`,

It could also be `pe_group['Survived'].apply(lambda x: x.sum () / x.count()) ` (for the sake of `apply lambda` tutorial)

````python
class_survivability = pe_group['Survived'].apply(lambda x: x.sum() / x.count()*100).reset_index()
class_survivability
````

If we plot it out,

````python
sns.barplot(x = "Pclass", y="Survived", hue="Embarked", data=class_survivability)
````

Last but not least, we can use all of the aggregation function in single line by using `.agg()` function.

````python
df.groupby('Embarked')['Fare'].agg(['min', 'max', 'mean', 'median', 'sum', 'count'])
````

# Combining Data

When performing operations on data, sometimes we have to combine different `DataFrames` or `Series`.

Pandas have 3 methods to combine them, `pd.concat()`, `DataFrame.join()`, and `pd.merge()`, but we will skip `DataFrame.join()` because we are able to do same thing with `pd.merge()`.

**pd.concat()**

This function concatenate list of pandas objects along a particular axis. So, it's basically combine data horizontally or vertically.

Let's try to concatenate these 2 following DataFrame horizontally.

**pd.merge()**

`pd.merge()` merges DataFrame or Series with a database-style join. The join is done on columns or indexes.

4 basic arguments pd.merge() takes are:
* `left` : Left DataFrame
* `right` : Right DataFrame
* `how` : 'left', 'right', 'outer', 'inner'. Join method. default: 'inner'
* `on` : column name or list of column names to merge based on.


Suppose we want to obtain `Average Fare of Pclass Group` for every passengers and assign it into new column.

````python
avg_fare = df.groupby('Pclass')['Fare'].mean().reset_index().rename(columns={'Fare':'AVG_Fare'})
avg_fare
````

````python
df = pd.merge(df, avg_fare, how='left', on=['Pclass'])
df.head(5)
````

# Imputing Missing Values

Missing values are given the value `NaN` (Not a Number) and have `float64 dtype`. Handling those missing data often is necessary because they can cause problems due to incompatibility. On this section, we will learn how to handle them. This process is called `Imputation`.

First, let's check which column has missing values.

````python
df.isnull().sum()
````

- `Age` column has 177 missing values.
- `Cabin` column has 687 missing values.
- `Embarked` column has 2 missing values.

Now we will try to impute the missing values of `Age` column using `median` values based on their `sexes`.

````python
male_filler_age = df.loc[df['Sex']=='male', 'Age'].median()
female_filler_age = df.loc[df['Sex']=='female', 'Age'].median()

print('Male Filler Age:', male_filler_age)
print('Female Filler Age:', female_filler_age)
````

Now we will use those values to fill the missing values using `.loc` function.

````python
df.loc[(df['Age'].isnull()) & (df['Sex']=='male'), 'Age'] = male_filler_age
df.loc[(df['Age'].isnull()) & (df['Sex']=='female'), 'Age'] = female_filler_age
````

Then check it using `.isnull()`

````python
df[df['Age'].isnull()]
````

Pandas also provided `.fillna()` function to impute the NaN values.

Next we will impute `Cabin` column using `.fillna()` function with value `'Unknown'`.

Assigning NaN with values like `Unknown`, `Others`, `-1` is useful to prevent incompatibility issue without dismiss the missing meaning.

````python
df['Cabin'].head(5)
````

````python
df['Cabin'].fillna('Unknown', inplace=True)
````

````python
df['Cabin'].head(5)
````

Next, we will examine the meaning of `NaN` in `Embarked` column.

````python
df[df['Embarked'].isnull()]
````

Both of them is `Survived` the event. This give us a hint that they might be didn't get on Titanic. If that's the case, those are `outliers` in our data.

In statistics, `Outlier` is a data point that differs significantly from other observations and not relevant with others. Outlier may alters the statistics of our data. Therefore, we should drop them out.

To drop rows or columns with `Null` values, we can use `.dropna()` function.

````python
# subset, columns or rows that will be considered with default value of None (consider all of them).
df.dropna(subset=['Embarked'], inplace=True)
````

Now if we check with `.isnull().sum()` there will be no missing value anymore.

````python
df.isnull().sum()
````

## Covariance

Covariance is a measure of the joint variability of two random variables.

The Covariance value can be:

**Positive Covariance**: if the greater values of one variable mainly correspond with the greater values of the other variable, and the same holds for the lesser values, (i.e., the variables tend to show similar behavior). 

**Negative Covariance**: in the opposite case, when the greater values of one variable mainly correspond to the lesser values of the other, (i.e., the variables tend to show opposite behavior).

The sign of the covariance shows the tendency in the linear relationship between the variables. The magnitude of the covariance is not easy to interpret. The normalized version of the covariance, the correlation coefficient, however, shows by its magnitude the strength of the linear relation.

````python
df.cov()
````

## Correlation

Correlation coefficient is a bivariate analysis that measures the strength of association between two variables and the direction of the relationship.

There's 3 method provided, pearson, kendall, and spearman.

`.corr(method='pearson')`

`method = 'pearson', 'kendall', 'spearman'`

**Pearson** : measure the degree of the linear relationship.

**Kendall** : measures the strength of dependence between two variables. 

**Spearman**: measure the degree of association between two variables. 

````python
df.corr(method='pearson')
````
