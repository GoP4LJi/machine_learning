what is NumPy?
NumPy is the foundational Python library for numerical computing. Its central object is the homogeneous, multidimensional ndarray, which enables compact data storage and fast vectorized operations. This notebook explains the most useful functions with small, runnable examples.

```python
import numpy as np
# Q1: Create a NumPy array containing numbers from 1 to 10.

arr = np.arange(1, 11)
arr
```

```python
# Q2: Create an array containing even numbers from 2 to 20.
arr_even = np.arange(2, 21, 2)
arr_even
```

```python
# Q3: Create an array of 10 zeros.
arr_zeros = np.zeros(10, dtype=int)
arr_zeros
```

```python
# Q4: Create an array of 15 ones.
arr_ones = np.ones(15, dtype=int)
arr_ones
```

```python
# Q5: Create a 4×4 identity matrix.
arr = np.eye(4)
arr
```

```python
# Q6: Create a 3×5 array filled with the number 8.
arr = np.ones((3, 5), dtype=int) * 8
arr
```

```python
# Q7: Create an array containing 20 equally spaced values between 5 and 50.
arr = np.linspace(5, 50, 20 ,dtype=int)
arr
```

```python
# Q8: Create a random 3×4 array of floating-point numbers.
arr = np.random.rand(3, 4)
arr
```

```python
# Q9: Create a random 4×4 array of integers between 10 and 100.
arr = np.random.randint(10, 100, size = (4, 4))
arr
```

```python
# Q10: Print the following properties of an array:
# Shape
# Size
# Data type
# Number of dimensions

arr = np.array([[1, 2, 3], [4, 5, 6]])
print("Shape:", arr.shape)
print("Size:", arr.size)
print("Data type:", arr.dtype)
print("Number of dimensions:", arr.ndim)

```

```python
# Q11: Reshape a 1D array containing numbers from 1 to 12 into a 3×4 matrix.
arr = np.arange(1, 13).reshape(3, 4) # This reshapes the 1D array into a 3x4 matrix
arr
```

```python
# Q12: Convert a 4×5 matrix into a 1D array.
arr = np.random.randint(1, 100, size=(4, 5))
arr = arr.flatten() # This flattens the 4x5 matrix into a 1D array
arr
```

```python
# Q13: Access:
# First element
# Last element
# Middle element
# of a NumPy array.
arr = np.array([1, 2, 3, 4, 5])
print("First element:", arr[0])  # First element
print("Last element:", arr[-1])  # Last element
print("Middle element:", arr[len(arr)//2])  # Middle element  
```

```python
# Q14: Extract the first three rows from a 5×5 matrix.
arr = np.random.randint(1, 100, size=(5, 5))
print("Original array:\n", arr)
arr1 = arr[:3, :]  # Extracting the first three rows
arr1
```

```python
# Q15: Extract the last two columns from a matrix.
arr = np.random.randint(1, 100, size=(3, 3))
print("Original array:\n", arr)
arr1 = arr[:, -2:]  # Extracting the last two columns
arr1
```

```python
# Q16: Replace all values greater than 20 with 100.
arr = np.random.randint(1, 100, size=(3, 3))
print("Original array:\n", arr)
arr[arr > 20] = 100
print("Array after replacement:\n", arr)
```

```python
# Q17: Count the number of elements greater than 50.
arr = np.random.randint(1, 100, size=(3, 3))
print("Original array:\n", arr)
count = np.sum(arr > 50)
print("Number of elements greater than 50:", count)
```

```python
# Q18: Create a Boolean mask for even numbers.
arr = np.random.randint(1, 100, size = (3, 3))
print("Original array:\n", arr)
mask = arr % 2 == 0
print("Boolean mask for even numbers:\n", mask)
```

```python
# Q19: Print only the odd numbers from an array.
arr = np.random.randint(1, 100, size=(3, 3))
print("Original array:\n", arr)
odd_numbers = arr[arr % 2 != 0]
print("Odd numbers from the array:\n", odd_numbers)
```

```python
# Q20: Reverse a NumPy array.
arr = np.random.randint(1, 100, size=(3, 3))
print("Original array:\n", arr)
reversed_arr = arr[::-1, ::-1]  # Reversing the array
print("Reversed array:\n", reversed_arr)
```
