```python
# Custom Function for Min Max Scaling
def min_max_scaling(X):
    x_min=X.min()
    x_max=X.max()
    return (X-x_min)/(x_max-x_min)
```

# Mean Scaling using Custom Function

```python

def mean_scaling(X):
    x_mean=X.mean()
    x_max=X.max()
    x_min=X.min()
    return (X-x_mean)/(x_max-x_min)
```
