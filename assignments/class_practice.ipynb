```python
df.columns = df.columns.str.strip().str.lower()
```

```python
df.columns = df.columns.str.replace(r'[^a-zA-Z0-9]+', '_', regex=True).str.strip('_')
```

```python
df.columns
```

```python
df['hourly_rate_usd'] = pd.to_numeric(
    df['hourly_rate_usd'],
    errors='coerce'
)
```

```python
df['client_satisfaction'] = (
    df['client_satisfaction']
    .astype(str)
    .str.replace('%', '', regex=False)
)
```

```python
df['client_satisfaction'] = pd.to_numeric(
    df['client_satisfaction'],
    errors='coerce'
)
```

```python
df['age'] = df['age'].fillna(df['age'].median())
```

```python
df['years_of_experience'] = df['years_of_experience'].fillna(
    df['years_of_experience'].median()
)
```

```python
df['hourly_rate_usd'] = df['hourly_rate_usd'].fillna(
    df['hourly_rate_usd'].median()
)
```

```python
df['rating'] = df['rating'].fillna(
    df['rating'].median()
)
```

```python
df['client_satisfaction'] = df['client_satisfaction'].fillna(
    df['client_satisfaction'].median()
)
```

```python
df['is_active'] = df['is_active'].astype(str).str.strip().str.lower()
```

```python
df['is_active'] = df['is_active'].replace({
    'y': 1,
    'yes': 1,
    'true': 1,
    '1': 1,
    'n': 0,
    'no': 0,
    'false': 0,
    '0': 0,
    'nan': np.nan
})
```

```python
df['is_active'] = df['is_active'].fillna(
    df['is_active'].mode()[0]
)
```
